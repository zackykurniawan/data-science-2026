### Python — List Operations

### Python — Membuat NumPy Array

```python
import numpy as np

# Dari list Python
a1D = np.array([1, 2, 3, 4, 5]) # 1D
a2D = np.array([[1,2,3],[4,5,6]]) # 2D
# Array dengan nilai inisial
zeros = np.zeros((3, 4)) # nol semua
ones = np.ones((2, 3)) # satu semua
eye = np.eye(4) # matriks identitas
full = np.full((2, 2), 7) # nilai konstan
# Sequence
arange = np.arange(0, 20, 2) # 0,2,4,...18
linsp = np.linspace(0, 1, 11) # 11 titik merata

# Array acak (reproducible)
rng = np.random.default_rng(seed=42)
rand_int = rng.integers(0, 100, (3, 4))
rand_flt = rng.normal(loc=0, scale=1, size=(100,))

# Tampilkan semua
for name, arr in {
    "a1D": a1D, "a2D": a2D, "zeros": zeros,
    "ones": ones, "eye": eye, "full": full,
    "arange": arange, "linsp": linsp,
    "rand_int": rand_int
}.items():
    print(f"\n{name} {arr.shape}:\n{arr}")
```

### Python — NumPy Vektorisasi & Statistik

```python
# Operasi aritmetika vektorisasi
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])
print(a + b) # → [11 22 33 44 55]
print(a * b) # → [10 40 90 160 250]
print(a ** 2) # → [1 4 9 16 25]
# Universal functions (ufuncs)
np.sqrt(a) # akar kuadrat tiap elemen
np.exp(a) # eksponensial (e^x)
np.log(b) # logaritma natural
# Fungsi statistik
data = rng.normal(0, 1, 1000)
print(f'Mean: {np.mean(data):.4f}')
print(f'Std: {np.std(data):.4f}')
print(f'Min: {np.min(data):.4f}')
print(f'Max: {np.max(data):.4f}')
print(f'Q1: {np.percentile(data,25):.4f}')
print(f'Q3: {np.percentile(data,75):.4f}')
```

### Q2 — Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?

```python
missing_count = df.isnull().sum()
missing_pct   = (missing_count / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Jumlah Missing': missing_count,
    'Proporsi (%)'  : missing_pct
})

# Tampilkan hanya kolom yang ada missing value
missing_df = missing_df[missing_df['Jumlah Missing'] > 0].sort_values(
    'Jumlah Missing', ascending=False
)
print('Kolom dengan missing values:')
display(missing_df)
```

**Temuan Q2:**  
Terdapat tiga kolom dengan missing values: **`Cabin`** memiliki missing values terbesar (~77%), artinya sebagian besar penumpang tidak tercatat nomor kabinnya. Kolom **`Age`** memiliki sekitar 19% data kosong, sehingga perlu strategi imputasi jika akan dianalisis lebih lanjut. Kolom **`Embarked`** hanya memiliki 2 nilai kosong yang dapat diimputasi dengan modus.

### Q3 — Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas.

```python
# Cara 1: menggunakan NumPy
survived_array = df['Survived'].values          # konversi ke NumPy array
pct_numpy = np.mean(survived_array) * 100
print(f'[NumPy]  Penumpang selamat: {pct_numpy:.2f}%')

# Cara 2: menggunakan Pandas
pct_pandas = df['Survived'].mean() * 100
print(f'[Pandas] Penumpang selamat: {pct_pandas:.2f}%')

# Ringkasan jumlah
jumlah_selamat    = df['Survived'].sum()
jumlah_tidak      = len(df) - jumlah_selamat
print(f'\nSelamat     : {jumlah_selamat} penumpang')
print(f'Tidak selamat: {jumlah_tidak} penumpang')
```

**Temuan Q3:**  
Hanya sekitar **38.38%** penumpang yang selamat (342 dari 891 orang). Artinya lebih dari 60% penumpang meninggal dalam tragedi tenggelamnya kapal Titanic. Angka ini mencerminkan betapa sulitnya kondisi evakuasi saat itu.

### Q4 — Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata-rata usianya?

```python
# Boolean indexing: wanita DAN kelas 1
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

jumlah  = len(wanita_kelas1)
rata_usia = wanita_kelas1['Age'].mean()

print(f'Jumlah penumpang wanita kelas 1 : {jumlah}')
print(f'Rata-rata usia                  : {rata_usia:.2f} tahun')

# Tampilkan beberapa baris
display(wanita_kelas1[['Name', 'Age', 'Survived', 'Fare']].head(10))
```

**Temuan Q4:**  
Terdapat **94 penumpang wanita di kelas 1** dengan rata-rata usia sekitar **34–37 tahun**. Kelompok ini dikenal sebagai salah satu yang paling tinggi tingkat keselamatannya karena prioritas evakuasi diberikan kepada wanita dan anak-anak, khususnya dari kelas atas.

### Q5 — Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?

```python
# Groupby: tingkat keselamatan per kelas
survival_by_class = df.groupby('Pclass').agg(
    total_penumpang = ('PassengerId', 'count'),
    jumlah_selamat  = ('Survived', 'sum'),
    tingkat_selamat = ('Survived', 'mean'),
    rata_usia       = ('Age', 'mean'),
    rata_fare       = ('Fare', 'mean')
).round(3)

survival_by_class['tingkat_selamat_%'] = (
    survival_by_class['tingkat_selamat'] * 100
).round(2)

print('Tingkat Keselamatan per Kelas Penumpang:')
display(survival_by_class)

# Identifikasi kelas terbaik
kelas_terbaik = survival_by_class['tingkat_selamat'].idxmax()
print(f'\nKelas dengan tingkat keselamatan tertinggi: Kelas {kelas_terbaik}')
```

**Temuan Q5:**  
**Kelas 1** memiliki tingkat keselamatan tertinggi (~63%), diikuti Kelas 2 (~47%), dan Kelas 3 (~24%). Terdapat kesenjangan yang signifikan antara kelas 1 dan kelas 3, yang mencerminkan ketidaksetaraan akses terhadap sekoci penyelamat. Penumpang kelas 3 yang umumnya berada di dek bawah mengalami kesulitan lebih besar untuk mencapai titik evakuasi.

### Q6 — Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.

```python
# Analisis 1: Tingkat keselamatan berdasarkan jenis kelamin
print('=== Tingkat Keselamatan per Jenis Kelamin ===')
survival_sex = df.groupby('Sex').agg(
    total           = ('PassengerId', 'count'),
    selamat         = ('Survived', 'sum'),
    tingkat_selamat = ('Survived', 'mean')
).round(3)
survival_sex['persen_%'] = (survival_sex['tingkat_selamat'] * 100).round(2)
display(survival_sex)

# Analisis 2: Pivot table — Jenis kelamin vs Kelas
print('\n=== Tingkat Keselamatan: Jenis Kelamin × Kelas (%) ===')
pivot = df.groupby(['Pclass', 'Sex'])['Survived'] \
           .mean().unstack().round(3) * 100
pivot.columns.name = None
pivot.index.name   = 'Kelas'
display(pivot)

# Analisis 3: Distribusi usia penumpang yang selamat vs tidak
print('\n=== Distribusi Usia: Selamat vs Tidak ===')
usia_stats = df.groupby('Survived')['Age'].describe().round(2)
usia_stats.index = ['Tidak Selamat (0)', 'Selamat (1)']
display(usia_stats)

# Analisis 4: Proporsi penumpang anak-anak (< 18 tahun) yang selamat
df['Kategori_Usia'] = pd.cut(
    df['Age'],
    bins=[0, 17, 35, 60, 100],
    labels=['Anak (<18)', 'Dewasa Muda (18-35)',
            'Dewasa (36-60)', 'Lansia (>60)']
)
print('\n=== Tingkat Keselamatan per Kategori Usia ===')
survival_age = df.groupby('Kategori_Usia', observed=True)['Survived'] \
                  .agg(['count', 'sum', 'mean']).round(3)
survival_age.columns = ['Total', 'Selamat', 'Tingkat Keselamatan']
survival_age['Persen (%)'] = (survival_age['Tingkat Keselamatan'] * 100).round(2)
display(survival_age)
```

**Temuan Q6:**

**A. Jenis Kelamin vs Keselamatan:**  
Wanita memiliki tingkat keselamatan yang jauh lebih tinggi (~74%) dibandingkan pria (~19%). Kebijakan *"women and children first"* terbukti sangat berpengaruh dalam proses evakuasi.

**B. Kombinasi Kelas & Jenis Kelamin:**  
Wanita kelas 1 dan kelas 2 memiliki tingkat keselamatan sangat tinggi (>85%), sedangkan pria kelas 3 hanya sekitar 13%. Ini menunjukkan bahwa status sosial-ekonomi (tercermin dari kelas tiket) bersama jenis kelamin merupakan faktor penentu keselamatan yang sangat kuat.

**C. Usia & Keselamatan:**  
Anak-anak (<18 tahun) memiliki tingkat keselamatan yang lebih baik dibandingkan dewasa, konsisten dengan kebijakan prioritas evakuasi. Sebaliknya, lansia (>60 tahun) cenderung memiliki tingkat keselamatan yang lebih rendah, kemungkinan karena keterbatasan mobilitas.

**Kesimpulan:** Tiga faktor utama yang menentukan peluang selamat di Titanic adalah **jenis kelamin**, **kelas tiket**, dan **usia** — dengan wanita muda dari kelas atas memiliki peluang tertinggi.
